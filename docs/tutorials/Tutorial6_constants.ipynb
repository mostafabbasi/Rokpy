# **Tutorial 6:** Constants
This module contain a set of common data containers (dataclasses) and reference tables of constants. 

## Data Classes (Data Containers)

Each *row* or field in `rokpy` tables is a data-class. In python, data-classes are data containers that used to group a set of related properties together.

Here is the list of data classes in `constants` module and their corresponding key parameters:

- `ElasticPropertySet`: Stores *density*, *bulk* and *shear* properties
- Empirical Coefficients:
    - `GardnerCoeffcients`: Stores Gardner's *multiplier* and *exponent* coefficients
    - `CastagnaCoefficients`: Stores Castagna's parabolic *coeffcients*
    - `VernikCoeffcients`: Stores Vernik's squared parabolic *coeffcients*
    - `HanCoeffcients`: Stores Han's *intercept* and porosity and clay content *coefficients*
- `PropertyTemplate`: Store the *type*, *unit*, *plot range*, *symbol* and name *aliases* of a property, e.g. P-velocity

These are a set of examples showing how to define and use these data classes:

```python
from rokpy.constants import ElasticPropertySet, PropertyTemplate, HanCoefficients, PropertyNames

quartz_properties = ElasticPropertySet(density=2.65, bulk=36, shear=45)
print(f'Custom quartz Properties:\n{quartz_properties}\n')

density_template = PropertyTemplate(type=PropertyNames.Rho, unit="g/cc", plot_range=(1.45, 2.95), symbol="rho", alias_list=["RHOB", "rho", "density", "den"])
print(f'Custom density template:\n{density_template}\n')

shalysand_vp_han_coefs = HanCoefficients(intercept=5900, phi_coef=-9400, clay_coef=-2250, type="ShalySand")
print(f'ShalySand P-wave Han coefficients:\n{shalysand_vp_han_coefs}\n')
```

## Tables of Constants

Tables of constants are used to store a set of predefined data-classes (e.g. `MineralsPropertyTable` stores the elastic properties of different mineralogies). Users may refer to these tables to get a set of standard values rather than defining new ones. For example, instead of defining a new quartz property set by manually assigning density, bulk and shear moduli, one can access `MineralsPropertyTable` to get the standard values for this mineral.

`rokpy` includes following tables in `constants` module:

- `PropertyTemplates`
- `MineralsPropertyTable` (See more in [This](./Tutorial1a_RockComponentsModeling.ipynb#Table-of-Minerals-Property) Tutorial.)
- `FluidsPropertyTable` (See more in [This](./Tutorial1a_RockComponentsModeling.ipynb#Table-of-Fluids-Property) Tutorial.)

and Empirical coefficients tables (See more in [This](./Tutorial5_EmpiricalMethods.ipynb#Tables-of-Standard-Coefficients) Tutorial.)

- `GardnerTable`
- `CastagnaTable`
- `HanTable`
- `VernikTable`

User have to instantiate an object of these tables prior to be able to access the contents. Once the table object is instatiated, user may modify, remove or add new elements to the table. This makes it possible to customize the table or even create several objects for different scenarios or different intervals.

```python
from rokpy.constants import MineralsPropertyTable, ElasticPropertySet

# Instantiate an object of MineralsPropertyTable
minerals_table = MineralsPropertyTable()

print(f'Standard quartz Properties from MineralsPropertyTable:\n{minerals_table.Quartz}\n')

# Modify the bulk modulus of quartz in the minerals table
minerals_table.Quartz.bulk = 38
print(f'Modified quartz Properties from MineralsPropertyTable:\n{minerals_table.Quartz}\n')

# Add new mineral properties (Note the first item in the table)
minerals_table.add_item('CustomQuartz', ElasticPropertySet(density=2.65, bulk=37, shear=45))
print(f'Customized MineralsPropertyTable:\n{minerals_table}\n')
```

## Calculate Fluid Properties

In addition to the predefined properties for different pore fluids, users may also calculate the brine, oil, gas and CO₂ properties under in-situ conditions. `rokpy` use the Batzle and Wang (1992) and Xu (2006) relations to calculate these properties. These relations are wrapped into `BatzleWang` class (in *fluid_properties* module). `FluidsPropertyTable` internally use these relations to update its properties as follows:

```python
from rokpy.constants import FluidsPropertyTable

# Instantiate an object of FluidsPropertyTable
fluids_table = FluidsPropertyTable()
print(f'Default brine Properties from FluidsPropertyTable:\n{fluids_table.Brine}\n')

# Calculate and update fluids property table
fluids_table.calculate_brine(T=80., P=20., salinity=70000.)
print(f'Updated brine Properties:\n{fluids_table.Brine}\n')
```
